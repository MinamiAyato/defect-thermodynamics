# MgO Defect Thermodynamics from Scratch

## Milestone 1: Host Structure Preparation

### Research Question

How can an MgO crystal structure be prepared as a reliable host for bulk and defect calculations?

### Physics

A crystal structure contains:

- lattice vectors;
- atomic species;
- atomic coordinates;
- crystal symmetry.

Primitive and conventional cells represent the same crystal.

They contain different numbers of atoms.

A reliable host structure is required before:

- convergence testing;
- bulk relaxation;
- defect generation;
- VASP calculations.

### Workflow

$$
\text{crystallographic data}
\rightarrow
\text{pymatgen Structure}
\rightarrow
\text{cell standardisation}
\rightarrow
\text{bulk POSCAR}
$$

### Goals

1. Create the MgO host as a `pymatgen` `Structure`.
2. Inspect its composition, lattice, volume, and symmetry.
3. Generate the primitive and conventional standard cells.
4. Select and document the host cell used for later calculations.
5. Write the selected structure as a VASP POSCAR file.
6. Validate the written structure by reading it back into Python.

```python
## Exercise 1.1 — Create the MgO Host Structure

### Purpose

Create the MgO host structure from its crystallographic parameters.

### Input

- Crystal structure: rocksalt
- Space group: Fm-3m
- Lattice parameter: 4.212 Å
- Mg fractional coordinate: (0, 0, 0)
- O fractional coordinate: (0.5, 0.5, 0.5)

### Expected output

A `pymatgen` `Structure` object containing the complete MgO unit cell.
```

```python
from pathlib import Path
from pymatgen.core import Structure
```

```python
notebook_dir = Path.cwd()
project_root = notebook_dir.parent

data_root = (
    project_root
    / "data"
    / "raw"
    / "mgo_defect_thermodynamics_dataset"
)

bulk_relaxation_dir = (
    data_root
    / "02_bulk_relaxed"
)

initial_structure_path = (
    bulk_relaxation_dir
    / "POSCAR"
)

relaxed_structure_path = (
    bulk_relaxation_dir
    / "CONTCAR"
)

print("Initial structure:", initial_structure_path)
print("Initial file exists:", initial_structure_path.exists())

print("Relaxed structure:", relaxed_structure_path)
print("Relaxed file exists:", relaxed_structure_path.exists())
```

```python
#Load the initial structure
initial_bulk = Structure.from_file(initial_structure_path)
print("Initial object type:", type(initial_bulk))
```

```python
#Load the relaxed structure
relaxed_bulk = Structure.from_file(relaxed_structure_path)
print("Relaxed object type:", type(relaxed_bulk))
```

```python
#Get some basic information about the structure
print("Initial composition:", initial_bulk.composition)
print("Initial number of sites:", initial_bulk.num_sites)
print("Initial lattice lengths:", initial_bulk.lattice.abc)
print("Initial lattice angles:", initial_bulk.lattice.angles)
print("Initial volume:", initial_bulk.volume)
print("Initial space group:", initial_bulk.get_space_group_info())

print()

print("Relaxed composition:", relaxed_bulk.composition)
print("Relaxed number of sites:", relaxed_bulk.num_sites)
print("Relaxed lattice lengths:", relaxed_bulk.lattice.abc)
print("Relaxed lattice angles:", relaxed_bulk.lattice.angles)
print("Relaxed volume:", relaxed_bulk.volume)
print("Relaxed space group:", relaxed_bulk.get_space_group_info())
```

## Exercise 1.2 — Extract the Primitive Structure

### Purpose

Extract the primitive MgO structure from the relaxed bulk supercell.

The primitive structure will be used as the host input for `doped`.

```python
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer
```

```python
symmetry_analyzer = SpacegroupAnalyzer(relaxed_bulk)
relaxed_primitive = symmetry_analyzer.get_primitive_standard_structure()

print("Composition:", relaxed_primitive.composition)
print("Reduced formula:", relaxed_primitive.composition.reduced_formula)
print("Number of sites:", relaxed_primitive.num_sites)
print("Lattice lengths:", relaxed_primitive.lattice.abc)
print("Lattice angles:", relaxed_primitive.lattice.angles)
print("Volume:", relaxed_primitive.volume)
print("Space group:", relaxed_primitive.get_space_group_info())
```

```python
size_ratio = relaxed_bulk.num_sites / relaxed_primitive.num_sites
print("Size ratio (bulk/primitive):", size_ratio)
```

```python
# Compare the original and primitive structures
print("Original composition:", mgo_entry.structure.composition)
print("Primitive composition:", prim_struc.composition)

print("Original number of sites:", mgo_entry.structure.num_sites)
print("Primitive number of sites:", prim_struc.num_sites)

print("Original lattice lengths:", mgo_entry.structure.lattice.abc)
print("Primitive lattice lengths:", prim_struc.lattice.abc)

print("Original lattice angles:", mgo_entry.structure.lattice.angles)
print("Primitive lattice angles:", prim_struc.lattice.angles)
```

## Exercise 1.4 — Write and Validate the Initial POSCAR

### Purpose

- Write the primitive structure to a POSCAR file.
- Confirm that the file was created.
- Read the POSCAR back into Python.
- Verify that the structure was preserved.

```python
from pathlib import Path

project_root = Path.cwd().parent

output_directory = project_root / "MgO" / "Input_files"

output_directory.mkdir(
    parents=True,
    exist_ok=True,
)

poscar_output_path = output_directory / "prim_struc_POSCAR"

prim_struc.to(
    filename= poscar_output_path
)

print("Output path:", poscar_output_path)
print("File exists:", poscar_output_path.exists())
```

```python
from pymatgen.core import Structure
written_structure = Structure.from_file(poscar_output_path)

print("Original composition:", prim_struc.composition)
print("Written composition:", written_structure.composition)

print("Original sites:", prim_struc.num_sites)
print("Written sites:", written_structure.num_sites)

print("Structures equal:", written_structure == prim_struc)
```
