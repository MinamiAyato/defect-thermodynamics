# MgO Defect Thermodynamics from Scratch

## Module 3 — Defect Generation with doped

### Research Question

Which intrinsic defects and charge states can occur in MgO?

### Workflow

The relaxed primitive structure is passed to `DefectsGenerator`.

`doped` identifies symmetry-distinct defects, constructs a suitable supercell and proposes charge states.

### Physics

A defect is simulated periodically in a finite supercell.

The supercell should be large enough to reduce interactions between periodic defect images. Charge states describe the exchange of electrons with the electron reservoir.

### Goals

1. Initialize `DefectsGenerator`.
2. Inspect the generated defect types.
3. Locate the Mg_O antisite.
4. Inspect its supercell and composition.
5. Inspect its predicted charge states.
6. Write the generated defect data.
